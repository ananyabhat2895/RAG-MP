# Medicinal Plants RAG

This notebook inspects the existing ChromaDB persistence and runs retrieval against the current plant chunks without regenerating embeddings.


## Section 1 — Environment and imports


```python
from pathlib import Path
import json
import time
import re

import numpy as np
import pandas as pd

import ollama
import chromadb
from rank_bm25 import BM25Okapi

from medicinal_plants_rag import (
    BM25Retriever,
    EMBEDDING_MODEL,
    GENERATION_MODEL,
    TOP_K_VECTOR,
    TOP_K_BM25,
    TOP_K_FINAL,
    build_condition_context,
    build_context,
    build_plant_catalog,
    classify_intent,
    detect_condition,
    ensure_model,
    find_chroma_db_path,
    find_collection_name,
    generate_answer,
    list_models,
    query_embedding,
    reciprocal_rank_fusion,
    resolve_plant,
    retrieve,
    vector_search,
    check_ollama,
)

PROJECT_ROOT = Path.cwd().resolve().parent
print('PROJECT_ROOT=', PROJECT_ROOT)
```

## Section 2 — Configuration


```python
EMBEDDING_MODEL = 'qwen3-embedding:0.6b'
GENERATION_MODEL = 'qwen2.5:3b'
TOP_K_VECTOR = 10
TOP_K_BM25 = 10
TOP_K_FINAL = 5
RRF_K = 60
OLLAMA_URL = 'http://localhost:11434'

DB_PATH = find_chroma_db_path(PROJECT_ROOT)
COLLECTION_NAME = find_collection_name(DB_PATH)
print('DB_PATH=', DB_PATH)
print('COLLECTION_NAME=', COLLECTION_NAME)
```

## Section 3 — Ollama connection


```python
print('Ollama running:', check_ollama())
models = list_models()
print('Installed models:', models)
for model in [EMBEDDING_MODEL, GENERATION_MODEL]:
    if model not in models:
        print(f'Pulling missing model: {model}')
        ensure_model(model)

embedding_test = ollama.embed(model=EMBEDDING_MODEL, input='embedding test')
print('Embedding dimension:', len(embedding_test['embeddings'][0]))
start = time.time()
answer = ollama.chat(model=GENERATION_MODEL, messages=[{'role': 'user', 'content': 'Say hello in one sentence.'}], options={'temperature': 0.1, 'num_predict': 32})
print('Generation latency seconds:', round(time.time() - start, 3))
print(answer['message']['content'])
```

## Section 4 — ChromaDB connection


```python
client = chromadb.PersistentClient(path=str(DB_PATH))
collection = client.get_collection(name=COLLECTION_NAME)
count = collection.count()
print('Collection count:', count)
sample = collection.get(limit=3, include=['documents', 'metadatas'])
print('Sample IDs:', sample['ids'])
print('Sample metadata keys:', list(sample['metadatas'][0].keys())[:10])
print('Sample document prefix:', sample['documents'][0][:400])
try:
    collection_metadata = collection.metadata
    print('Collection metadata:', collection_metadata)
except Exception:
    print('No collection metadata available.')
```

## Section 5 — Plant catalog and alias resolution


```python
catalog = build_plant_catalog(collection)
print('Catalog size:', len(catalog))
tests = ['Neem', 'Tulsi', 'Ashwagandha', 'Azadirachta indica', 'Withania somnifera']
for test in tests:
    result = resolve_plant(test, catalog)
    print(test, '->', result)
```

## Section 6 — Vector retrieval


```python
question = 'What are the medicinal uses of Neem?'
resolved = resolve_plant(question, catalog)
if resolved:
    plant_id = resolved.get('plant_id')
else:
    plant_id = None
vector_hits = vector_search(collection, question, plant_id=plant_id, top_k=TOP_K_VECTOR)
print('Vector hits:', len(vector_hits))
for item in vector_hits[:3]:
    print(item['chunk_id'], item['plant_id'], item['metadata'].get('section'))
```

## Section 7 — BM25 retrieval


```python
bm25 = BM25Retriever(collection)
bm25_hits = bm25.search(question, plant_id=plant_id, top_k=TOP_K_BM25)
print('BM25 hits:', len(bm25_hits))
for item in bm25_hits[:3]:
    print(item['chunk_id'], item['plant_id'], round(float(item['score']), 3))
```

## Section 8 — Reciprocal rank fusion


```python
fused = reciprocal_rank_fusion([vector_hits, bm25_hits], k=RRF_K)
print('Fused results:', len(fused))
for item in fused[:3]:
    print(item['chunk_id'], round(float(item['score']), 5))
```

## Section 9 — Intent classification


```python
tests = [
    'What are the medicinal uses of Neem?',
    'What is the botanical name of Neem?',
    'What is the family of Neem?',
    'What are the vernacular names of Neem?',
    'Where is Neem distributed?',
]
for test in tests:
    print(test, '->', classify_intent(test))
```

## Section 10 — Unified retrieval and answer generation


```python
questions = [
    'What is the botanical name of Neem?',
    'What is the family of Neem?',
    'What are the medicinal uses of Neem?',
    'Tell me about Tulsi.',
    'Tell me about Unknownplant.'
]
for question in questions:
    start = time.time()
    resolved = resolve_plant(question, catalog)
    plant_id = resolved.get('plant_id') if resolved else None
    ranked = retrieve(question, collection, bm25, top_k=TOP_K_FINAL, plant=plant_id)
    context = build_context(ranked)
    answer = generate_answer(question, context)
    print('QUESTION:', question)
    print('RESOLVED:', resolved)
    print('MATCH_COUNT:', len(ranked))
    print('ANSWER:', answer[:500])
    print('SECONDS:', round(time.time() - start, 3))
    print('---')
```

## Section 11 — Condition support


```python
condition_context = build_condition_context()
print(condition_context)
for question in ['Which plants are associated with fever?', 'Is Neem associated with fever?']:
    print(question, '->', detect_condition(question))
```

## Section 12 — Final summary


```python
print('RAG PIPELINE STATUS')
print('Ollama:', check_ollama())
print('Embedding Model:', EMBEDDING_MODEL)
print('Generation Model:', GENERATION_MODEL)
print('ChromaDB:', DB_PATH)
print('Total Chunks:', collection.count())
print('Plant Resolver:', 'available' if catalog else 'missing')
print('Vector Search:', 'available')
print('BM25:', 'available')
print('RRF:', 'available')
print('Plant RAG:', 'available')
```
